```python
# Pie chart for race/ethnicity distribution (RIDRETH3)
import matplotlib.pyplot as plt

# Using combined_df and RIDRETH3 column
race_counts = combined_df['RIDRETH3'].value_counts().sort_index()

# Map codes to descriptions as per your provided list
race_labels = {
    1: 'Mexican American',
    2: 'Other Hispanic',
    3: 'Non-Hispanic White',
    4: 'Non-Hispanic Black',
    5: 'Non-Hispanic Asian',  # Note: Your list has 6 for Asian, but data uses 5
    6: 'Non-Hispanic Asian',  # Matching your list
    7: 'Other Race - Including Multi-Racial'  # As per your list (not "Refused")
}

# Prepare labels and values
labels = [race_labels.get(code, f'Code {code}') for code in race_counts.index]
sizes = race_counts.values

# Create pie chart
fig, ax = plt.subplots(figsize=(10, 8))
ax.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=plt.cm.tab10.colors)
ax.set_title('Race/Ethnicity Distribution (RIDRETH3)', fontsize=14, fontweight='bold')
ax.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.

plt.tight_layout()
plt.savefig('race_ethnicity_pie_chart.png', dpi=300, bbox_inches='tight')
plt.show()

print("Pie chart saved as 'race_ethnicity_pie_chart.png'")
print(f"Total participants: {race_counts.sum()}")

# Display percentage table
print("\n" + "="*60)
print("RACE/ETHNICITY DISTRIBUTION TABLE")
print("="*60)
print(f"{'Code':<5} {'Description':<35} {'Count':>8} {'Percentage':>10}")
print("-"*60)
for code in race_counts.index:
    desc = race_labels.get(code, f'Code {code}')
    count = race_counts[code]
    pct = (count / race_counts.sum()) * 100
    print(f"{code:<5} {desc:<35} {count:>8,} {pct:>9.2f}%")
print("="*60)
```
